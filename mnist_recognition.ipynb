```python
from sklearn import datasets
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split

digits = datasets.load_digits()

X = digits.data
Y = digits.target

scaler = MinMaxScaler()
X_normalized_minmax = scaler.fit_transform(X)

train_data, test_data, train_target, test_target = train_test_split(
    X_normalized_minmax,
    Y,
    test_size=0.2,
    shuffle=True,
    random_state=42,
)

image_size = len(train_data[0])
target_size = 10
```

```python
import numpy as np


def get_label(digits_array):
    labels = []
    for digit in digits_array:
        label = np.zeros((10))
        label[digit] = 1
        labels.append(label)
    return labels


train_labels = get_label(train_target)
test_labels = get_label(test_target)
```

```python
import neuralnetwork

nn = neuralnetwork.NeuralNetwork(
    input_size=image_size,
    hidden_size=64,
    output_size=target_size,
)

nn.fit(
    train_data,
    train_labels,
    epochs=100,
    lr=0.9,
    momentum_beta=0.9,
    dorpout_p=0.8
)
```

```python
nn._accuracy(test_data, test_labels)
```
